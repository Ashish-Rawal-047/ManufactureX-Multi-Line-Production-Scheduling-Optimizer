# 04 — SimPy Discrete-Event Simulation & Monte Carlo Robustness Analysis

Evaluates deterministic optimized schedules under stochastic processing times, arrival delays, and random breakdowns.

```python
import sys
from pathlib import Path
sys.path.insert(0, str(Path('..').resolve()))

import pandas as pd
from plantsched.baselines import solve_fcfs
from plantsched.core import CPSATConfig, ObjectiveWeights, SimulationConfig
from plantsched.data import InstanceGeneratorConfig, generate_instance
from plantsched.optimization import solve_cp_sat
from plantsched.simulation import run_monte_carlo

problem = generate_instance(InstanceGeneratorConfig(num_jobs=25, num_machines=3, num_products=6, seed=42))
weights = ObjectiveWeights()
fcfs = solve_fcfs(problem, weights=weights)
cpsat = solve_cp_sat(problem, CPSATConfig(max_time_seconds=10.0, weights=weights))

sim_cfg = SimulationConfig(processing_time_cv=0.12, num_replications=30, random_seed=42)
mc_fcfs = run_monte_carlo(problem, fcfs, sim_cfg)
mc_cp = run_monte_carlo(problem, cpsat, sim_cfg)
pd.DataFrame([mc_fcfs.to_summary_dict(), mc_cp.to_summary_dict()])
```
