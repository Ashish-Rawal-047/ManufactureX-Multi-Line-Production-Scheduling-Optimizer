# 02 — Exact Optimization: FCFS vs. CP-SAT vs. MILP

Compares First-Come First-Served (FCFS), Google OR-Tools CP-SAT, and PuLP MILP on an identical manufacturing instance.

```python
import sys
from pathlib import Path
sys.path.insert(0, str(Path('..').resolve()))

from plantsched.core import CPSATConfig, MILPConfig, ObjectiveWeights
from plantsched.data import InstanceGeneratorConfig, generate_instance
from plantsched.evaluation import compare_algorithms
from plantsched.visualization import plot_gantt_matplotlib

problem = generate_instance(InstanceGeneratorConfig(num_jobs=15, num_machines=3, num_products=5, seed=42))
weights = ObjectiveWeights()
comp = compare_algorithms(
    problem,
    algorithms=('FCFS', 'CP-SAT', 'MILP'),
    weights=weights,
    cp_sat_config=CPSATConfig(max_time_seconds=15.0, weights=weights),
    milp_config=MILPConfig(max_time_seconds=15.0, weights=weights),
)
comp.summary_table
```

```python
fig = plot_gantt_matplotlib(problem, comp.schedules['CP-SAT'])
fig
```
