# 03 — Metaheuristics: Genetic Algorithm & Simulated Annealing

Evaluates GA and SA on larger scheduling instances and inspects convergence curves.

```python
import sys
from pathlib import Path
sys.path.insert(0, str(Path('..').resolve()))

from plantsched.core import GAConfig, ObjectiveWeights, SAConfig
from plantsched.data import InstanceGeneratorConfig, generate_instance
from plantsched.evaluation import compare_algorithms

problem = generate_instance(InstanceGeneratorConfig(num_jobs=100, num_machines=4, num_products=8, seed=101))
weights = ObjectiveWeights()
comp = compare_algorithms(
    problem,
    algorithms=('FCFS', 'GA', 'SA'),
    weights=weights,
    ga_config=GAConfig(population_size=50, generations=80, random_seed=42, weights=weights),
    sa_config=SAConfig(max_iterations=1500, random_seed=42, weights=weights),
)
comp.summary_table
```
